```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model = "gemini-3.1-flash-lite"
)
```

```python
def print_response(response):
    for block in response.content_blocks:
        if block['type'] == 'text':
            print(block['text'])
```

## Zero shot prompting



```python
response = llm.invoke("What is inertia?")
print_response(response)
```

## Few shot prompts

```python
from langchain.messages import HumanMessage, SystemMessage, AIMessage
```

```python
messages = [
    HumanMessage("what is 2 * 2 + 3 ?"),
    AIMessage("""
    step 1: 2 * 2 = 4
    step 2: 4 + 3 = 7

    Answer: 7
    """),
    HumanMessage("What is 3 - 3 + 6 + 4 ?"),
    AIMessage("""
    step 1: 3 - 3 = 0
    step 2: 0 + 6 = 6
    step 3: 6 + 4 = 10

    Answer: 10
    """),
    HumanMessage("what is 5 * 6 - 2 ?")
    
]
```

```python
response = llm.invoke(messages)
print_response(response)
```

## Creating models just by passing a string

* Convention `provider:model`

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
# approach 1
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

response = model.invoke("What is capital of France?")
response.pretty_print()



```

```python
# approach 2

model = init_chat_model(
    model="gemini-3.5-flash-lite",
    model_provider="google_genai"
)
response = model.invoke("What is capital of France?")
response.pretty_print()
```

```python
# approach 3
model = init_chat_model(model="google_genai:gemini-3.5-flash-lite")
response = model.invoke("What is capital of France?")
response.pretty_print()
```

```python
# creating agent
from langchain.agents import create_agent
# approach 1
# create model
model = init_chat_model(model="google_genai:gemini-3.5-flash-lite")
agent = create_agent(
    model=model,
    tools = []
)
response = agent.invoke({
    "messages": ["What is capital of france"]
})
for message in response['messages']:
    print(message)
```

```python
# approach 2
agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools = []
)
response = agent.invoke({
    "messages": ["What is capital of france"]
})
for message in response['messages']:
    print(message)
```

```python
model = init_chat_model("openrouter:openrouter/free")
response = model.invoke("What is capital of France?")
response.pretty_print()
```

```python
model_name="z-ai/glm-5.2:free"
model_provider="openrouter"
model = init_chat_model(f"{model_provider}:{model_name}")
response = model.invoke("What is capital of France?")
response.pretty_print()
```

```python
def print_agent_response(response):
    for message in response['messages']:
        print(message)
```

```python
agent_without_shortterm_memory = create_agent(
    model='google_genai:gemini-3.5-flash-lite'
)

response = agent_without_shortterm_memory.invoke(
    {
        "messages": "My name is khaja"
    }
)
print_agent_response(response)

response = agent_without_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    }
)
print_agent_response(response)

```

```python
from langgraph.checkpoint.memory import InMemorySaver
cfg_1 = {"configurable": {"thread_id": "classroom-02sep"}}
cfg_2 = {"configurable": {"thread_id": "klsdjfdaslk"}}

agent_with_shortterm_memory = create_agent(
    model='google_genai:gemini-3.5-flash-lite',
    checkpointer=InMemorySaver(),
)

response = agent_with_shortterm_memory.invoke(
    {
        "messages": "My name is khaja"
    },
    cfg_1
)
print_agent_response(response)

```

```python
# agent with short term same thread/conversation
response = agent_with_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    },
    cfg_1
)
print_agent_response(response)

```

```python
# agent with short term different thread/conversation
response = agent_with_shortterm_memory.invoke(
    {
        "messages": "What is my name?"
    },
    cfg_2
)
print_agent_response(response)
```

## Longterm memory

```python
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()

store.put(('hr', 'hyd', 'abc'),'favorites', {'color': 'blue', 'food': 'biriyani'} )
store.put(('it', 'hyd', 'xyz'),'favorites', {'color': 'red', 'food': 'dosa'} )
```

```python
store.list_namespaces()
```

```python
store.get(('hr', 'hyd', 'abc'), 'favorites')
```

```python
store.search(
    ('hr', 'hyd', 'abc'),
    query='biryani'
)
```

```python
agent_with_shortterm_memory = create_agent(
    model='google_genai:gemini-3.5-flash-lite',
    checkpointer=InMemorySaver(), # short term memory
    store=store  # long term memory
)
```
