## Message in langchain or langgraph

```python
from langchain.chat_models import init_chat_model, BaseChatModel
from dotenv import load_dotenv
```

```python
model_name = 'google_genai:gemini-3.5-flash-lite'
load_dotenv()
```

```python
llm = init_chat_model(
    model_name
)
```

## Style 1

* Objects

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage
result = llm.invoke([
    SystemMessage(content="You are an helpful assistant"),
    HumanMessage(content="What is capital of France ?")
])
result.pretty_print()
```

```python
result = llm.invoke([
    SystemMessage(content="You are an helpful assistant", id="abc1"),
    HumanMessage(content="What is capital of France ?", id="abc2")
])
result.pretty_print()
```

## Style 2: Tuple

```python
result = llm.invoke([
    ("system", "You are an helpful assistant"),
    ("human", "What is capital of France?")
])

result.pretty_print()
```

## Style 3

* dictionary or openai style


```python
result = llm.invoke([
    {"role": "system", "content": "You are an helpful assistanct"},
    {"role": "human", "content": "What is capital of France?"}
])
result.pretty_print()
```

```python
result = llm.invoke([
    {"role": "system", "content": "You are an helpful assistanct", "id": "abc1"},
    {"role": "human", "content": "What is capital of France?", "id": "abc2"}
])
result.pretty_print()
```
